```python
import seaborn as sns
import matplotlib.pyplot as plt
data={
    "product":["laptop","phone","tablet","laptop","phone","tablet"],
    "sales":[50,80,40,70,90,60],
    "price":[50000,30000,20000,55000,32000,22000],
    "months":["jan","jan","feb","feb","mar","mar"]
}
sns.barplot(x="product",y="sales",data=data)
plt.show()
sns.scatterplot(x="product",y="sales",data=data)
plt.show()
sns.lineplot(x="product",y="sales",data=data)
plt.show()
sns.countplot(x="product",data=data)
plt.show()
sns.histplot(x="product",data=data)
plt.show()
sns.boxplot(x="product",y="sales",data=data)
plt.show()
sns.violinplot(x="product",y="sales",data=data)
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
data=np.random.randn(1000)
plt.figure(figsize=(8,6))
sns.histplot(data,bins=30,kde=True,color="yellow",edgecolor="black")#kde=kernel density distribution
plt.xlabel("value")
plt.ylabel("frequency")
plt.title("distribution ofn random data in seaborn")
plt.grid(axis="y",alpha=1)#alpha to change transparency of color according to value
plt.show()
```

```python
'''ques 1-online store sales
an online store has record the number of units sold for different product.the manager want to compare
the performance and easily identify which product is sold most.
task= visualise the data in a suitable way.'''
import matplotlib.pyplot as plt
products=["laptop","phone","tablet","watch"]
sales=[50,80,40,70]
plt.bar(products,sales)
plt.xlabel("products")
plt.ylabel("sales")
plt.title("online store sales")
plt.show()
```

```python
'''ques 2-website visitor
A company reacorded its website visitor during a working week,the marketing team wants  to understand how visitor
activity changed throughout the week.
Task=creat a suitable visualisation to show this change.'''
import matplotlib.pyplot as plt
days=["mon","tue","wed","thus","fri"]
visitors=[120,150,180,160,210]
plt.plot(days,visitors,marker="o",linestyle="--",color="red")
plt.xlabel("days")
plt.ylabel("visitors")
plt.title("website visitors")
plt.grid(True)
plt.show()
```

```python
'''ques 3- monthly expense
A student recorded their monthly expense.The student want to understand how the total spending is divided among differnt categories.
Task=represent the information visually'''
import matplotlib.pyplot as plt
categories=["food","travel","rent","shopping"]
expenses=[2000,1500,5000,3000]
plt.pie(expenses,labels=categories,autopct="%1.1f%%",startangle=90)
plt.axis("equal")
plt.title("monthly expenses")
plt.show()
```

```python
'''ques 4- Advertising and sales
An e-commerce company recorded its advertising expenditure and sales.
The company want to investigate whether changes in advertising expenditure are associated with changes in sales.
Task= create a suitable visualization for this analysis'''
import matplotlib.pyplot as plt
advertising=[10,20,30,40,50]
sales=[25,35,45,55,70]
plt.scatter(advertising,sales)
plt.xlabel("advertising")
plt.ylabel("sales")
plt.title("advertising,sales")
plt.show()
```

```python
'''ques 5- Student marks
A teacher has collected the marks of a class.
The teacher wants to  understand how the marks are distributed among the students.
Task=visualise the distriburtion of marks.'''
import matplotlib.pyplot as plt
marks=[45,50,52,60,65,70,
       72,75,78,80,82,85,88,90,95]
plt.hist(marks,bins=5,edgecolor="black")
plt.xlabel("marks")
plt.ylabel("frequency")
plt.title("student marks")
plt.show()
```
